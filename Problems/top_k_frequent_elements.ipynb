```python
Given an integer array nums and an integer k, return the k most frequent elements. 
You may return the answer in any order.

 

Example 1:

Input: nums = [1,1,1,2,2,3], k = 2

Output: [1,2]

Example 2:

Input: nums = [1], k = 1

Output: [1]

Example 3:

Input: nums = [1,2,1,2,1,2,3,1,3,2], k = 2

Output: [1,2]

 

Constraints:

1 <= nums.length <= 105
-104 <= nums[i] <= 104
k is in the range [1, the number of unique elements in the array].
It is guaranteed that the answer is unique.
 

Follow up: Your algorithm's time complexity must be better than O(n log n), where n is the array's size.
```

```python
class Solution(object):

    def topKFrequent(self, nums, k):

        count = {}

        for num in nums:
            count[num] = 1 + count.get(num, 0)

        arr = []

        for num, cnt in count.items():
            arr.append([cnt, num])

        arr.sort()

        res = []

        while len(res) < k:
            res.append(arr.pop()[1])

        return res
```

```python
class Solution(object):

    def topKFrequent(self, nums, k):

        count = {}

        # Count each number
        for num in nums:
            if num not in count:
                count[num] = 1
            else:
                count[num] += 1

        # Store [frequency, number]
        arr = []

        for num in count:
            arr.append([count[num], num])

        # Sort by frequency
        arr.sort()

        # Get top k
        res = []

        while len(res) < k:
            item = arr.pop()
            res.append(item[1])

        return res
```

```python
The solution first counts how many times each number appears using the count dictionary.
 Then it stores each number with its frequency in arr as [frequency, number]. 
 After sorting arr, the most frequent numbers are at the end, so pop() takes them one
  by one and adds them to res until we have k numbers. Finally, res is returned.
 Time: O(n + m log m), Space: O(m).
```

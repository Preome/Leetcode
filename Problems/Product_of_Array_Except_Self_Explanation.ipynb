238. Product of Array Except Self - Explanation
Problem Link

Description
Given an integer array nums, return an array output where output[i] is the product of all the elements of nums except nums[i].

Each product is guaranteed to fit in a 32-bit integer.

Follow-up: Could you solve it in 
O
(
n
)
O(n) time without using the division operation?

Example 1:

Input: nums = [1,2,4,6]

Output: [48,24,12,8]
Example 2:

Input: nums = [-1,0,1,2,3]

Output: [0,-6,0,0,0]
Constraints:

2 <= nums.length <= 100,000
-30 <= nums[i] <= 30
The product of any prefix or suffix of nums is guaranteed to fit in a 32-bit integer.

```python
class Solution(object):

    def productExceptSelf(self, nums):

        n = len(nums)
        res = [0] * n

        for i in range(n):

            prod = 1

            for j in range(n):

                if i == j:
                    continue

                prod *= nums[j]

            res[i] = prod

        return res
```

For every position i, we calculate the product of all elements except nums[i]. prod starts at 1, then the inner loop multiplies every other number. The result is stored in res[i]. For example, [1,2,3,4] gives [24,12,8,6].

Time: O(n²)
Space: O(n) for the result array.

```python
class Solution(object):

    def productExceptSelf(self, nums):

        prod = 1
        zero_cnt = 0

        # Find product and count zeros
        for num in nums:
            if num != 0:
                prod *= num
            else:
                zero_cnt += 1

        # More than one zero
        if zero_cnt > 1:
            return [0] * len(nums)

        res = [0] * len(nums)

        # Calculate result
        for i in range(len(nums)):

            if zero_cnt > 0:
                if nums[i] == 0:
                    res[i] = prod
                else:
                    res[i] = 0

            else:
                res[i] = prod // nums[i]

        return res
```

First, we calculate the product of all non-zero numbers and count how many zeros exist. If there are 2 or more zeros, every answer is 0. If there is exactly 1 zero, only the position containing zero gets the product; all other positions get 0. If there are no zeros, we divide the total product by the current number to get the product of everything except that number.

Time: O(n)
Space: O(n) for the result array.

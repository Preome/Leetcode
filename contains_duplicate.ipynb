```python
Given an integer array nums, return true if any value appears more than once in the array,
otherwise return false.

Example 1:

Input: nums = [1, 2, 3, 3]

Output: true

Example 2:

Input: nums = [1, 2, 3, 4]

Output: false
Constraints:

0 <= nums.length <= 10^5
-10^9 <= nums[i] <= 10^9
```

```python
class Solution(object):

    def containsDuplicate(self, nums):

        for i in range(len(nums)):
            for j in range(i + 1, len(nums)):
                if nums[i] == nums[j]:
                    return True

        return False


O(n^2)
```

```python
class Solution(object):

    def containsDuplicate(self, nums):

        seen = set()

        for num in nums:
            if num in seen:
                return True

            seen.add(num)

        return False


O(n)
```
